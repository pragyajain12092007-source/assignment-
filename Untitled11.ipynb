```python
import numpy as np
# Part 1: Create Marks Matrix
rng = np.random.default_rng()
marks = rng.integers(0, 101, size=(10, 5))
print(marks)

# Part 2: Calculate Total Marks and Percentage
total = np.sum(marks, axis=1)
print(total)
percentage = (total / 500) * 100
print(percentage)

# Part 3
average = np.mean(marks, axis=0)
print(average)
std_dev = np.std(marks, axis=0)
print(std_dev)
highest_marks = np.max(marks, axis=0)
print(highest_marks)
topper_index = np.argmax(marks, axis=0)
print(topper_index)

# Part 4: Sort Marks Using argsort()
sorted_indices = np.argsort(marks, axis=0)
print(sorted_indices)

# Part 5: Identify Pass and Fail Using Boolean Mask
passing_marks = 33
pass_mask = marks >= passing_marks
print("\nPass/Fail in Each Subject (True = Pass, False = Fail):")
print(pass_mask)
student_pass = np.all(pass_mask, axis=1)
print("\nResult of Each Student:")
print(student_pass)

```
